## Set up steps (you only need to do this once):

1. If you are off campus, login to the UiB VPN (Cisco).  

2. Open the terminal. 

    **Windows**: On windows 10, you can open the terminal by writing 'terminal' in the windows search bar. If your windows version is norwegian, you may also search for 'Ledertekst'. Finally, if you have anaconda installed on your personal computer, you may also use the windows search bar to find the 'anaconda prompt'. These are all different ways to access the terminal.

    **Linux or Mac**: open the terminal app.

3. In the terminal, log in to the remote machine called 'cyclone' by entering the command written below. Make sure to change the username to your UiB user name and press enter. Then enter your UiB password and press enter.

```
ssh your.user.name@cyclone1.gfi.uib.no
```

4. In the terminal, load the shared Anaconda module. This gives you Python, JupyterLab, and most of the packages we need, without installing anything yourself:

```
module load Anaconda3
```

5. In the terminal, download the python materials for the course by entering the command below. Then cd into the course directory folder.

```
git clone https://github.com/Emil-Bruvik/geof212/ ~/geof212
```
```
cd ~/geof212/
```

6. In the terminal, install the couple of packages the Anaconda module doesn't already include (`cartopy` and `netCDF4`). `--user` installs them into your own home directory, not into the shared module, so nothing already provided gets re-downloaded:

```
pip install --user -r requirements.txt
```

7. In the terminal, start Jupyter lab by entering the command below. Some urls should pop up in the terminal. 

    **It's important that you choose a 4-digit number that is different from the one provided (8080). Only one person at a time can use a given port number and several of you will be using jupyter lab at once.** 

```
jupyter-lab --no-browser --ip="cyclone1.gfi.uib.no" --port=8080
```

8. Open up a browser and enter one of the URLs given from step 7. Voila! Now you're ready to code in python on a remote supercomputer using the jupyter notebook on your internet browser.  

    To play around with the jupyter notebook, take a look at the notebook for this exercise by opening the file: 
    
    ```
    ~/geof212/python.exercises/00.python.exercise.ipynb
    ```
